# Estrutura de Repetição (Loop)

## - For 


```python
for x in range(5):
    print(x)
```

```python
compras=["uva","laranja","ovos"]
for x in compras:
    print(f"comprei {x}")
```

## - For  - Compreention


```python
l=[1,2,3,4,12,13,21]
pares=[p for p in l if p%2==0]
print(pares)
```

## - While

```python
l=[1,2,3,4,12,13,21]
pares = []
cont=0
while (cont < len(l)):
    if l[cont]%2==0:
        pares.append(l[cont])
    cont+=1
print(cont)
```
